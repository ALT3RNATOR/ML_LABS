```python
import numpy as np
import matplotlib.pyplot as plt

# Generate linear data
np.random.seed(0)

x = 2 * np.random.rand(100, 1)
y = 4 + 3 * x + np.random.randn(100, 1)

X = np.c_[np.ones((len(x), 1)), x]
m = len(y)

# Gradient descent
theta = np.zeros((2, 1))

for _ in range(100):
    error = X @ theta - y
    theta -= 0.1 * (2 / m) * X.T @ error

# Closed-form solution
theta_closed = np.linalg.inv(X.T @ X) @ X.T @ y

# Calculate cost
theta0 = np.linspace(-2, 6, 100)
theta1 = np.linspace(0, 5, 100)

t0, t1 = np.meshgrid(theta0, theta1)

J = np.array([
    [
        np.mean(
            (X @ np.array([[a], [b]]) - y) ** 2
        )
        for a in theta0
    ]
    for b in theta1
])

# Plot cost surface
fig = plt.figure()

ax = fig.add_subplot(
    111,
    projection="3d"
)

ax.plot_surface(t0, t1, J)
plt.show()

# Plot contours
plt.contour(t0, t1, J)
plt.scatter(theta[0], theta[1])
plt.show()

print("Theta (Gradient Descent):\n", theta)
print("Theta (Closed Form):\n", theta_closed)
```
