```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix

# Load dataset
data = load_breast_cancer(as_frame=True)
df = data.frame

print("Shape:", df.shape)

print(
    "\nMissing Values:\n",
    df.isnull().sum()
)

print(
    "\nData Types:\n",
    df.dtypes
)

print(
    "\nZero Values:\n",
    (df == 0).sum()
)

print("\nMean Radius:", df["mean radius"].mean())

# Split selected features and target
X = df[
    [
        "mean radius",
        "mean texture",
        "mean perimeter",
        "mean area",
        "mean smoothness"
    ]
]

y = df["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=0
)

# Confusion Matrix
y_pred = (y_test == y_test)  # Placeholder removed below

from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(random_state=0)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

tn, fp, fn, tp = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1]
).ravel()

print("\nConfusion Matrix:")
print([[tn, fp], [fn, tp]])

# Evaluation metrics
print("\nAccuracy:", (tp + tn) / (tp + tn + fp + fn))

print("Precision:", tp / (tp + fp) if tp + fp else 0)

print("Recall:", tp / (tp + fn) if tp + fn else 0)
```
